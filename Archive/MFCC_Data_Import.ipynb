```python
#summary: Function to extract the ID, take, and label from the filename
#summary: this is used for parsing the file name into individual variables
def extract_id_take_label(filename):
    import re
    match = re.match(r"(\d+)-(\d+)-(\w+)-(\w+).(\w+)", filename)
    #print(match)
    if match:
        return match
    return None

#Summary: Edition for augmented data
def extract_id_take_label_augmented(filename):
    import re
    match = re.match(r"(\d+)-(\d+)-(\w+)-(\w+)-(\w+).(\w+)", filename)
    #print(match)
    if match:
        return match
    return None
```

```python
def augmentation(data,AUG_type=None):
    match AUG_type:
        case "pitch" | 0:
            sr  = 16000
            bins_per_octave = 12
            pitch_pm = 2
            pitch_change =  pitch_pm * 2*(np.random.uniform())   
            audio_pitchy = librosa.effects.pitch_shift(data.astype('float64'),  sr, n_steps=pitch_change, bins_per_octave=bins_per_octave)
            return audio_pitchy
        
        case "noise" | 1:
            noise = np.random.normal(0, 0.1, len(data))
            audio_noisy = data + noise
            return audio_noisy

        case "rnd_shift" | 2:
            timeshift_fac = 0.2 *2*(np.random.uniform()-0.5)  # up to 20% of length
            start = int(data.shape[0] * timeshift_fac)
            if (start > 0):
                data = np.pad(data,(start,0),mode='constant')[0:data.shape[0]]
            else:
                data = np.pad(data,(0,-start),mode='constant')[0:data.shape[0]]
            return data
        case "amp_scale" | 3:
            sr  = 16000
            dyn_change = np.random.uniform(low=1.5,high=2.5)
            data = data * dyn_change
            return data
        case "time_stretch" | 4:
            rate=1.5
            input_length = len(data)
            streching = data.copy()
            streching = librosa.effects.time_stretch(streching, rate=rate)
            
            if len(streching) > input_length:
                streching = streching[:input_length]
            else:
                streching = np.pad(data, (0, max(0, input_length - len(streching))), "constant")
            return streching
        case None:
            pass


    
def audio_augmentation(path,AUG_type=None):
    sample_rate=0
    aug_data=0
    if AUG_type is None:
            data, sample_rate = librosa.load(path) #sr=None <- true sample rate
            aug_data = data
    else:
            data, sample_rate = librosa.load(path,sr=16000) #sr=None <- true sample rate
            aug_data = augmentation(data,AUG_type)
            #aug_data = np.array(aug_data,dtype='float32').reshape(-1,1)
        
    return compute_log_mel_spectogram(aug_data,sample_rate)

test = audio_augmentation("/Users/daniel/Desktop/Github/ESC_Dissertation_2024_EEE/ESC-50-master/Audio/1-137-A-32.wav",AUG_type = "noise")
print(test.shape)

data, sample_rate = librosa.load("/Users/daniel/Desktop/Github/ESC_Dissertation_2024_EEE/ESC-50-master/Audio/1-137-A-32.wav") #sr=None <- true sample rate
lms = compute_log_mel_spectogram(data,sample_rate)
#print(lms.shape)
#print(data.shape)
# 
# noise = np.random.normal(0, 0.1, len(data))
# audio_noisy = data + noise
# aug = np.array(audio_noisy,dtype='float32').reshape(-1,1)
# print(f"{data.shape}\t{audio_noisy.shape}\t{aug.shape}")
```

```python
import torchvision.transforms as T
import librosa
import numpy as np
import torch



def compute_log_mel_spectogram(audio,sample_rate):  # Size of the Fast Fourier Transform (FFT), which will also be used as the window length
    

    n_fft=1024
    hop_length=512
    window_type ='hann'
    mel_bins = 60 # Number of mel bands
    normalized_y = librosa.util.normalize(audio)
    Mel_spectrogram = librosa.feature.melspectrogram(y=normalized_y, sr=sample_rate,hop_length=hop_length, win_length=n_fft, n_mels = mel_bins)
    
    mel_spectrogram_db = librosa.power_to_db(Mel_spectrogram)
    delta=librosa.feature.delta(mel_spectrogram_db)
    
    #stack data together
    data = np.dstack( (mel_spectrogram_db,delta) ) #print(data.shape)
    data_tensor = torch.tensor(data, dtype=torch.float32) #print(data_tensor.shape)
    data_tensor =  data_tensor.permute(2, 0, 1)##reshape\swap axis #print(data_tensor.shape)
    
    #DEBUG - checks that data is the same, the sum==0 means it has the same data
    #tmp = torch.load(destination_file)
    #torch.sum(torch.add(-tmp,data_tensor))

    #resize all data to the same size
    # Convert to a 4D tensor, assuming the data represents 1-channel images
    data_tensor = data_tensor.unsqueeze(1)  # [2, 1, 60, 288] - 1 channel added
    resize = T.Resize((60, 216))
    data_tensor_resized = resize(data_tensor)
    data_tensor_resized = data_tensor_resized.squeeze(1)  # Back to [2, 60, 216]
    
    #print(data_tensor_resized.shape)  # torch.Size([2, 60, 216])
    return  data_tensor_resized
```
